```python
import pandas as pd 
```

```python
pontos = [50, 1, 2, 100, 70, 23, 80]

pontos_50_mais = []

for i in pontos:
    if i >=50:
        pontos_50_mais.append(i)

print(pontos_50_mais)
        
```

```python
brinquedo = pd.DataFrame(
    {
        "Nome" :[ "Alison", "Ana", "José"],
        "Idade" : [21,20,16],
        "uf" : ["sp", "rj", "pr"]
    }
)

brinquedo
```

### Filtro

```python
filtro = brinquedo["Idade"] >=18
```

```python
brinquedo[filtro]
```

```python
filtro = df["QtdePontos"] >= 50 
```

```python
df[filtro].sample(5)
```

## Filtro com duas condições "AND = &"

```python
filtro = (df["QtdePontos"] >= 50) &  (df["QtdePontos"] <100)
```

```python
df[filtro].sample(5)
```

## Filtro operador "OR = |" 

```python
filtro = (df["QtdePontos"] == 1) | (df["QtdePontos"] == 100)  
```

```python
df[filtro].tail(6)
```
