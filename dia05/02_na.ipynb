# NaN

```python
import pandas as pd 
```

```python
clientes = pd.read_csv("../data/clientes.csv", sep=";")
```

#### ".dropna" Utilizado para deletar o row  Com o (how= "All") precisa q todos os registros tenha Nan
#### já com o (how= "any") apenas uma coluna para deletar a row

```python
clientes.dropna().head(5)
```

```python
df = pd.DataFrame(
    {

        "Nome" : ["Alison", None, "Ana","Marli"],
        "Idade" : [None, None, 20, 45],
        "Salario" : [ 2345, 3422, None, 6532]

    }
)
```

 #### "subsat = ["Coluna"] " para selecionar a coluna que deseja tirar o NaN

```python
df.dropna(how="all", subset=["Idade"])
```

## ".fillna()" para imputar um valor

```python
df.fillna("Não se aplica")
```

#### Para ser em df tem que atribuir para o Dataframe

```python
df_fill = df["Idade"].fillna("Não se aplica")
```

```python
df_fill.fillna("Não se aplica")
```

### Utilizado fill em formato de Dicionario podemos organizar os NaN por coluna 
###### Exemplo:

```python
df.fillna({"Nome" : "Não registrado", "Idade" : 0, "Salario" : "Nulo" })
```

## Media

```python
medias = df[["Idade", "Salario"]].mean()
```

```python
df.fillna(medias)
```
