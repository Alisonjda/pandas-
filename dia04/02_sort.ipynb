# Ordenando duas colunas 

```python
brinquedo = pd.DataFrame(
{
    "Nome" : ["Alison", "José", "Ana", "marli"],
    "Idade" : [23,18,23,45],
    "Salario" : [2345, 4533, 3245, 4533],
}
)

brinquedo
```

```python
brinquedo.sort_values(["Salario", "Idade"], ascending=[False, True])
```
